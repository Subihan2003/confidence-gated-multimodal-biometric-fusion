```python
import pandas as pd

base_path = "/content/drive/MyDrive/voxceleb/voxceleb_210/"

csv_files = [
    "train_metadata.csv",
    "train.csv",
    "val.csv",
    "test.csv"
]

prefix = "/content/drive/MyDrive/voxceleb/voxceleb_210/"

for file in csv_files:

    file_path = base_path + file

    df = pd.read_csv(file_path)

    df["path"] = df["path"].str.replace(
        prefix,
        "",
        regex=False
    )

    df.to_csv(file_path, index=False)

    print(f"{file} updated successfully.")
```
